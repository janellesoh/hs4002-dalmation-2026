```python
df['ppethm'] = df['ppethm'].str.lower()
df['pp2_ppethm'] = df['pp2_ppethm'].str.lower()
df['pp3_ppethm'] = df['pp3_ppethm'].str.lower()
```

```python
df['respondent_race_raw'] = df['ppethm'].fillna(df['pp2_ppethm']).fillna(df['pp3_ppethm'])
df['respondent_race_raw'].value_counts(dropna=False)
```

```python
race_map = {
    'white, non-hispanic': 'White',
    'hispanic': 'Hispanic',
    'black, non-hispanic': 'Black',
    '2+ races, non-hispanic': 'Other/Multiracial',
    'other, non-hispanic': 'Other/Multiracial'
}
df['respondent_race_group'] = df['respondent_race_raw'].map(race_map)
df['respondent_race_group'].value_counts(dropna=False)
```

```python
partner_race_map = {
    'white': 'White',
    'black or african american': 'Black',
    'asian or pacific islander': 'Other/Multiracial',
    'american indian, aleut, or eskimo': 'Other/Multiracial',
    'other (please specify)': 'Other/Multiracial',
    'refused': None
}
df['partner_race_group'] = df['q6b'].map(partner_race_map)
df['partner_race_group'].value_counts(dropna=False)
```

```python
df['respondent_race_group'].value_counts(dropna=False)
```

```python
race_table = df['respondent_race_group'].value_counts(dropna=False).reset_index()
race_table.columns = ['Race Group', 'Count']
race_table.to_csv('../output/tables/race_distribution.csv', index=False)
race_table
```
